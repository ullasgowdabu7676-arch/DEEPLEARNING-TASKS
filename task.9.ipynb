```python
# Task 9: Simple YOLO-style Loss

import torch
import torch.nn as nn

# Simple YOLO Loss

class YOLOLoss(nn.Module):

    def __init__(self):
        super(YOLOLoss, self).__init__()

        # Mean Squared Error for Bounding Box
        self.mse = nn.MSELoss()

        # Binary Cross Entropy for Classification
        self.bce = nn.BCELoss()

    def forward(self, pred_box,
                      true_box,
                      pred_class,
                      true_class):

        # Bounding Box Loss
        box_loss = self.mse(pred_box, true_box)

        # Classification Loss
        class_loss = self.bce(pred_class, true_class)

        # Total Loss
        total_loss = box_loss + class_loss

        return total_loss



# Predicted Bounding Box
# (x_center, y_center, width, height)
pred_box = torch.tensor([[0.45, 0.50, 0.30, 0.35]])

# Ground Truth Bounding Box
true_box = torch.tensor([[0.50, 0.55, 0.28, 0.40]])

# Predicted Class Probability
pred_class = torch.tensor([[0.80]])

# Actual Class
true_class = torch.tensor([[1.0]])

# Create Loss Object

loss_function = YOLOLoss()

loss = loss_function(
    pred_box,
    true_box,
    pred_class,
    true_class
)

print("YOLO Loss =", loss.item())
```
