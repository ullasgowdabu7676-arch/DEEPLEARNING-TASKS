```python
# Task 3: 2D Convolution from Scratch


import numpy as np

# Sample 5x5 Input Image
image = np.array([
    [1, 2, 3, 2, 1],
    [4, 5, 6, 5, 4],
    [7, 8, 9, 8, 7],
    [4, 5, 6, 5, 4],
    [1, 2, 3, 2, 1]
])

# 3x3 Filter (Kernel)
kernel = np.array([
    [1, 0, -1],
    [1, 0, -1],
    [1, 0, -1]
])

# Image and Kernel Size
image_size = image.shape[0]
kernel_size = kernel.shape[0]

# Output Size
output_size = image_size - kernel_size + 1

# Empty Output Matrix
output = np.zeros((output_size, output_size))

# Perform Convolution
for i in range(output_size):
    for j in range(output_size):
        region = image[i:i+kernel_size, j:j+kernel_size]
        output[i, j] = np.sum(region * kernel)

# Display Results
print("Input Image:\n")
print(image)

print("\nKernel:\n")
print(kernel)

print("\nConvolution Output:\n")
print(output)
```
