```python
# Task 4: Simple Optimizers

import numpy as np

weight = 5.0
learning_rate = 0.1
gradient = 2.0

print("Initial Weight:", weight)

# Gradient Descent
gd = weight - learning_rate * gradient
print("\nGradient Descent:", gd)

# Momentum
momentum = 0.9
velocity = 0
velocity = momentum * velocity - learning_rate * gradient
momentum_weight = weight + velocity
print("Momentum:", momentum_weight)

# RMSProp
beta = 0.9
cache = 0
cache = beta * cache + (1-beta) * gradient**2
rms = weight - learning_rate * gradient / (np.sqrt(cache)+1e-8)
print("RMSProp:", rms)

# Adam
m = 0
v = 0
beta1 = 0.9
beta2 = 0.999

m = beta1*m + (1-beta1)*gradient
v = beta2*v + (1-beta2)*(gradient**2)

m_hat = m/(1-beta1)
v_hat = v/(1-beta2)

adam = weight - learning_rate*m_hat/(np.sqrt(v_hat)+1e-8)

print("Adam:", adam)
```
