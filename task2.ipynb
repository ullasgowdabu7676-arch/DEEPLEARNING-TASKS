```python
import numpy as np

# Number of neurons
input_neurons = 4
output_neurons = 3

# Random Initialization
random_weights = np.random.randn(input_neurons, output_neurons)

# Xavier Initialization
xavier_weights = np.random.randn(input_neurons, output_neurons) * np.sqrt(1 / input_neurons)

# He Initialization
he_weights = np.random.randn(input_neurons, output_neurons) * np.sqrt(2 / input_neurons)

print("Random Initialization")
print(random_weights)

print("\nXavier Initialization")
print(xavier_weights)

print("\nHe Initialization")
print(he_weights)
```
