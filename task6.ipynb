```python
# Task 6: Simple BiLSTM Example

import torch
import torch.nn as nn


x = torch.randn(2, 5, 10)

# BiLSTM Model
class BiLSTM(nn.Module):
    def __init__(self):
        super(BiLSTM, self).__init__()

        self.lstm = nn.LSTM(
            input_size=10,
            hidden_size=16,
            num_layers=1,
            bidirectional=True,
            batch_first=True
        )

        self.fc = nn.Linear(32, 2)   # 16 x 2 directions

    def forward(self, x):

        output, (hidden, cell) = self.lstm(x)

        # Last Time Step
        out = output[:, -1, :]

        out = self.fc(out)

        return out


# Create Model
model = BiLSTM()

# Forward Pass
output = model(x)

print("Model Output:")
print(output)

print("\nOutput Shape:", output.shape)
```
